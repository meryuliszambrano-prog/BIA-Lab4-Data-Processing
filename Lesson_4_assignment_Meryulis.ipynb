## Task 3: Handling Null Values

```python
# Task 3: Handling Null Values
# Create a new DataFrame with some null values and fill them

# Your code here:
df_null = pd.DataFrame({
    'Product': ['Laptop', None, 'Mouse', 'Monitor'],
    'Price': [1200.0, 25.5, np.nan, 300.0],
    'Quantity': [5, np.nan, 10, np.nan]
})
print("Original DataFrame (at least one null in each column):")
print(df_null)
print("\nNull values per column:")
print(df_null.isnull().sum())
```

```python
# Method 1: fill with fixed values (0 for numeric columns, 'Unknown' for text)
filled = df_null.fillna({'Product': 'Unknown', 'Price': 0, 'Quantity': 0})
print("Method 1 - fill with 0 / 'Unknown':")
print(filled)

# Method 2: fill numeric columns with a computed value (mean / median)
filled_mean = df_null.copy()
filled_mean['Price'] = filled_mean['Price'].fillna(filled_mean['Price'].mean())
filled_mean['Quantity'] = filled_mean['Quantity'].fillna(filled_mean['Quantity'].median())
print("\nMethod 2 - fill with mean (Price) and median (Quantity):")
print(filled_mean)

# Method 3: forward fill (uses the previous value)
print("\nMethod 3 - forward fill:")
print(df_null.ffill())

# Method 4: drop the rows with nulls
print("\nMethod 4 - dropna (only complete rows are kept):")
print(df_null.dropna())
```

## Task 5: Concatenation and Merging

```python
# Task 5: Concatenation and Merging
# Concatenate two DataFrames vertically and merge them horizontally

df1 = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
df2 = pd.DataFrame({'A': [5, 6], 'B': [7, 8]})
df3 = pd.DataFrame({'C': [9, 10], 'D': [11, 12]})

# Your code here:
# 1. Vertical concatenation (axis=0); ignore_index creates a new index 0..3
df_vertical = pd.concat([df1, df2], axis=0, ignore_index=True)
print("df1 + df2 vertically:")
print(df_vertical, "\nShape:", df_vertical.shape)

# 2. Merge the result with df3 horizontally using the index
df_final = pd.merge(df_vertical, df3, left_index=True, right_index=True, how='left')
print("\nMerged with df3 horizontally:")
print(df_final, "\nShape:", df_final.shape)
```

```python
# Check: 4 rows and 4 columns, and all original values are present
print("Correct shape (4, 4)?", df_final.shape == (4, 4))
# df1 + df2 have 8 values and df3 has 4 values -> 12 non-null values expected
print("All 12 original values present?", df_final.notnull().sum().sum() == 12)
# df3 has only 2 rows, so rows 2 and 3 have NaN in C and D
```

## Task 6: Tuples and Sets

```python
# Task 6: Tuples and Sets
# Create a tuple of fruits and a set of numbers.
# Then, try to add an element to each and observe the difference.

# Your code here:
fruits = ('apple', 'banana', 'mango')
numbers_set = {1, 2, 3, 4, 5}
print("Tuple:", fruits)
print("Set:", numbers_set)

# Tuples are immutable: adding an element raises an error
try:
    fruits.append('orange')
except AttributeError as e:
    print("\nTuple error:", e)

# The only way is to create a NEW tuple
new_fruits = fruits + ('orange',)
print("New tuple:", new_fruits)

# Sets are mutable: add() works, but duplicates are ignored
numbers_set.add(6)
numbers_set.add(3)     # 3 already exists, the set does not change
print("\nSet after add(6) and add(3):", numbers_set)
```

**Difference:** a **tuple** is ordered and **immutable**: after it is created we cannot add, change or remove elements (it has no `append` or `add`). A **set** is **mutable** and unordered, and it only keeps **unique** values: `add()` inserts a new element and duplicates are ignored.

## Task 8: Functions and Lambda

```python
# Task 8: Functions and Lambda
# Create a function to calculate the square of a number.
# Then, use a lambda function to do the same.

# Your code here:
def square(x):
    """Return the square of a number. Raises an error if x is not a number."""
    if not isinstance(x, (int, float)):
        raise TypeError("x must be a number")
    return x ** 2

square_lambda = lambda x: x ** 2

for value in [3, -4, 2.5]:
    print(f"square({value}) = {square(value)} | lambda({value}) = {square_lambda(value)}")

try:
    square('five')
except TypeError as e:
    print("Error:", e)
```

## Task 9: Iterators and Generators

```python
# Task 9: Iterators and Generators
# Create an iterator for the first 5 even numbers.
# Then, create a generator for the same.

# Your code here:
class EvenIterator:
    """Iterator class that returns the first n even numbers (2, 4, 6, ...)."""
    def __init__(self, n=5):
        self.n = n
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.n:
            raise StopIteration
        self.count += 1
        return self.count * 2

it = EvenIterator(5)
print("Iterator with next():", next(it), next(it))
print("Rest of the iterator:", list(it))

def even_generator(n=5):
    """Generator function that yields the first n even numbers."""
    for i in range(1, n + 1):
        yield i * 2

gen = even_generator(5)
print("\nGenerator with next():", next(gen))
print("Rest of the generator:", list(gen))
print("Using both in a for loop:", [x for x in EvenIterator()], [x for x in even_generator()])
```

## Task 10: Map, Reduce, and Filter

```python
# Task 10: Map, Reduce, and Filter
# Use map to square all numbers in a list.
# Use reduce to find the product of all numbers in a list.
# Use filter to get only even numbers from a list.

numbers = [1, 2, 3, 4, 5]

# Your code here:
squared = list(map(lambda x: x ** 2, numbers))
print("map - squares:", squared)

# initial value 1 avoids an error if the list is empty
product = reduce(lambda a, b: a * b, numbers, 1)
print("reduce - product:", product)

evens = list(filter(lambda x: x % 2 == 0, numbers))
print("filter - even numbers:", evens)
```

## Task 11: Object-Oriented Programming - Creating a Class

```python
# Task 11: Object-Oriented Programming - Creating a Class
# Create a class called 'Rectangle' with attributes 'length' and 'width'.
# Include methods to calculate area and perimeter.

# Your code here:
class Rectangle:
    def __init__(self, length, width):
        # edge case: sides must be positive numbers
        if length <= 0 or width <= 0:
            raise ValueError("length and width must be positive")
        self.length = length
        self.width = width

    def area(self):
        return self.length * self.width

    def perimeter(self):
        return 2 * (self.length + self.width)

    def __str__(self):
        return "Rectangle %s x %s" % (self.length, self.width)

r1 = Rectangle(5, 3)
r2 = Rectangle(10, 4.5)
for r in [r1, r2]:
    print(r, "-> area:", r.area(), "| perimeter:", r.perimeter())

try:
    Rectangle(-2, 3)
except ValueError as e:
    print("Error:", e)
```
